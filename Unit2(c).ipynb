```python


import numpy as np
import matplotlib.pyplot as plt

L, N = 5.0, 1000
x = np.linspace(-L, L, N)
dx = x[1]-x[0]

# Kinetic energy matrix
main_diag = 2.0*np.ones(N)
off_diag = -1.0*np.ones(N-1)

T = (1.0/(2*dx**2))*(np.diag(main_diag)+np.diag(off_diag,1)+np.diag(off_diag,-1))

# Potential energy matrix
V_matrix = np.diag(0.5*x**2)
H = T + V_matrix

# Solve matrix
energies, wavefunctions = np.linalg.eigh(H)

plt.plot(x, 0.5*x**2, 'k-', linewidth=2, label='Potential V(x)')
print('---Energy Comparison---')

for n in range(3):
    E_ana = n+0.5
    print(f'State n={n}: Numerical E= {energies[n]:.4f} | Analytical E= {E_ana:.4f}')

    # Grab wavefunction column, normalize, and plot
    psi = wavefunctions[:,n]
    psi /= np.sqrt(np.sum(psi**2)*dx)

    plt.plot(x, energies[n]+psi, label=f'Num n={n}')

plt.ylim(0,3.5)
plt.xlim(-4,4)
plt.legend(loc='lower right')
plt.show()
```
